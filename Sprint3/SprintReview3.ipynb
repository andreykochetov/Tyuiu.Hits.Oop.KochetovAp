```

using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    protected Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo() => $"{Name} ({Price} руб.)";
    public override string ToString() => $"[{GetType().Name}] {GetInfo()}";

    public override bool Equals(object obj)
    {
        return obj is Component other && GetType() == other.GetType() && Name == other.Name && Price == other.Price;
    }
    public override int GetHashCode() => HashCode.Combine(GetType(), Name, Price);

    public static bool operator ==(Component left, Component right) => Equals(left, right);
    public static bool operator !=(Component left, Component right) => !Equals(left, right);
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public CPU(string name, decimal price, int cores) : base(name, price) => Cores = cores;
    
    public override string GetInfo() => base.GetInfo() + $" | Ядра: {Cores}";
    public void Install() => Console.WriteLine($"{Name} установлен в сокет.");
}

public class RAM : Component, IInstallable
{
    public int CapacityGB { get; set; }
    public RAM(string name, decimal price, int capacityGB) : base(name, price) => CapacityGB = capacityGB;
    
    public override string GetInfo() => base.GetInfo() + $" | Объем: {CapacityGB} ГБ";
    public void Install() => Console.WriteLine($"{Name} установлена в слот.");
}

public class GPU : Component, IInstallable
{
    public GPU(string name, decimal price) : base(name, price) { }
    public void Install() => Console.WriteLine($"{Name} установлена в PCI-E.");
}

public class HDD : Component, IInstallable
{
    public HDD(string name, decimal price) : base(name, price) { }
    public void Install() => Console.WriteLine($"{Name} подключен.");
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Remove(component)) Console.WriteLine($"{component.Name} удален.");
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var c in _components) total += c.Price;
        return total;
    }

    public void ShowInfo()
    {
        Console.WriteLine("\n--- ПК ---");
        foreach (var c in _components) Console.WriteLine(c.GetInfo());
        Console.WriteLine($"Итого: {GetTotalPrice()} руб.\n");
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC()
    {
        Computer pc = new Computer();
        pc.AddComponent(new CPU("Intel Core i9", 50000m, 8));
        pc.AddComponent(new RAM("Kingston 32GB", 10000m, 32));
        pc.AddComponent(new GPU("RTX 4080", 120000m));
        pc.AddComponent(new HDD("Samsung 1TB", 12000m));
        return pc;
    }
}

// --- Демонстрация ---
var cpu1 = new CPU("Intel i7", 30000m, 6);
var cpu2 = new CPU("Intel i7", 30000m, 6);
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 != cpu2: {cpu1 != cpu2}");

var pc = ComputerFactory.CreateGamingPC();
pc.ShowInfo();

pc.RemoveComponent(new HDD("Samsung 1TB", 12000m));
pc.ShowInfo();


```
