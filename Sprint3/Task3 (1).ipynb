<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет собой консольную модель CRM-системы, демонстрирующую ключевые принципы ООП в C#. В основе лежит абстрактный класс Person и базовый класс Customer, от которых наследуются три типа клиентов: VipCustomer, RegularCustomer и GroupCustomer. В коде реализованы инкапсуляция (свойства), наследование, интерфейсы (IIdentifiable, IContactable, ILoyalty), полиморфизм через переопределение (virtual/override) и перегрузку методов, а также генерический класс CustomerCollection<T> с ограничением where T : Customer для типизированного хранения объектов.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;

public interface IIdentifiable
{
    int CustomerId { get; set; }
    void ShowId();
}

public interface IContactable
{
    string Email { get; set; }
    string Phone { get; set; }
    void Contact();
}

public interface ILoyalty
{
    int LoyaltyPoints { get; set; }
    void AddPoints(int points);
}

public abstract class Person
{
    public string FirstName { get; set; }
    public string LastName { get; set; }
    public int Age { get; set; }
    public string Nationality { get; set; }

    protected Person(string firstName, string lastName, int age)
    {
        FirstName = firstName;
        LastName = lastName;
        Age = age;
        Nationality = "Не указано";
    }

    public virtual string GetFullName()
    {
        return $"{FirstName} {LastName}";
    }

    public virtual void Greet()
    {
        Console.WriteLine($"Привет, меня зовут {GetFullName()}");
    }

    public virtual void Greet(string greeting)
    {
        Console.WriteLine($"{greeting}, меня зовут {GetFullName()}");
    }
}

public class Customer : Person, IIdentifiable, IContactable
{
    public int CustomerId { get; set; }
    public string Email { get; set; }
    public string Phone { get; set; }
    public string Address { get; set; }
    public DateTime CreatedAt { get; set; }
    public bool IsActive { get; set; }
    public string PreferredLanguage { get; set; }

    public Customer(int customerId, string firstName, string lastName, int age,
                    string email, string phone, string address)
        : base(firstName, lastName, age)
    {
        CustomerId = customerId;
        Email = email;
        Phone = phone;
        Address = address;
        CreatedAt = DateTime.Now;
        IsActive = true;
        PreferredLanguage = "Русский";
    }

    public override string GetFullName()
    {
        return base.GetFullName();
    }

    public override void Greet()
    {
        base.Greet();
        Console.WriteLine($"Мой ID: {CustomerId}");
    }

    public override void Greet(string greeting)
    {
        base.Greet(greeting);
        Console.WriteLine($"Я клиент с ID: {CustomerId}");
    }

    public void ShowId()
    {
        Console.WriteLine($"ID клиента: {CustomerId}");
    }

    public void Contact()
    {
        Console.WriteLine($"Связь: {Email} | {Phone}");
    }

    public void Contact(string preferredMethod)
    {
        Console.WriteLine($"Связь через {preferredMethod}: {(preferredMethod == "Email" ? Email : Phone)}");
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"Email обновлён на {Email}");
    }

    public virtual void UpdateEmail(string newEmail, bool notify)
    {
        Email = newEmail;
        Console.WriteLine($"Email обновлён на {Email}. Уведомление: {(notify ? "Да" : "Нет")}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}, Имя: {GetFullName()}, Возраст: {Age}");
        Console.WriteLine($"Email: {Email}, Телефон: {Phone}, Адрес: {Address}");
    }

    public void Deactivate()
    {
        IsActive = false;
        Console.WriteLine($"Клиент {GetFullName()} деактивирован.");
    }
}

public class VipCustomer : Customer, ILoyalty
{
    public int LoyaltyPoints { get; set; }
    public string VipLevel { get; set; }
    public double DiscountPercent { get; set; }
    public DateTime VipSince { get; set; }
    public string PersonalManager { get; set; }

    public VipCustomer(int customerId, string firstName, string lastName, int age,
                       string email, string phone, string address,
                       int loyaltyPoints, string vipLevel, double discountPercent)
        : base(customerId, firstName, lastName, age, email, phone, address)
    {
        LoyaltyPoints = loyaltyPoints;
        VipLevel = vipLevel;
        DiscountPercent = discountPercent;
        VipSince = DateTime.Now;
        PersonalManager = "Не назначен";
    }

    public void AddPoints(int points)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"Начислено {points} баллов. Всего: {LoyaltyPoints}");
    }

    public void AddPoints(int points, string reason)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"Начислено {points} баллов за {reason}. Всего: {LoyaltyPoints}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"VIP-уровень: {VipLevel}, Баллы: {LoyaltyPoints}, Скидка: {DiscountPercent}%");
        Console.WriteLine($"VIP с: {VipSince:dd.MM.yyyy}, Персональный менеджер: {PersonalManager}");
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        Console.WriteLine($"VIP-клиент {GetFullName()} получил уведомление о смене email");
    }

    public void AssignPersonalManager(string managerName)
    {
        PersonalManager = managerName;
        Console.WriteLine($"Персональный менеджер {managerName} назначен для {GetFullName()}");
    }
}

public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }
    public int OrdersCount { get; set; }
    public decimal TotalSpent { get; set; }
    public bool IsSubscribed { get; set; }
    public string LoyaltyCardNumber { get; set; }

    public RegularCustomer(int customerId, string firstName, string lastName, int age,
                           string email, string phone, string address,
                           DateTime registrationDate, int ordersCount, decimal totalSpent)
        : base(customerId, firstName, lastName, age, email, phone, address)
    {
        RegistrationDate = registrationDate;
        OrdersCount = ordersCount;
        TotalSpent = totalSpent;
        IsSubscribed = true;
        LoyaltyCardNumber = "LC-" + customerId.ToString("D5");
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        Console.WriteLine($"Дата обновления: {DateTime.Now:dd.MM.yyyy HH:mm}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Дата регистрации: {RegistrationDate:dd.MM.yyyy}, Заказов: {OrdersCount}, Потрачено: {TotalSpent} руб.");
        Console.WriteLine($"Подписка: {(IsSubscribed ? "Активна" : "Отключена")}, Карта лояльности: {LoyaltyCardNumber}");
    }

    public void AddOrder(decimal amount)
    {
        OrdersCount++;
        TotalSpent += amount;
        Console.WriteLine($"Добавлен заказ на {amount} руб. Всего заказов: {OrdersCount}, сумма: {TotalSpent} руб.");
    }
}

public class GroupCustomer : Customer
{
    public string GroupName { get; set; }
    public int MembersCount { get; set; }
    public string CompanyType { get; set; }
    public decimal CorporateDiscount { get; set; }
    public string ContactPerson { get; set; }

    public GroupCustomer(int customerId, string firstName, string lastName, int age,
                         string email, string phone, string address,
                         string groupName, int membersCount, string companyType)
        : base(customerId, firstName, lastName, age, email, phone, address)
    {
        GroupName = groupName;
        MembersCount = membersCount;
        CompanyType = companyType;
        CorporateDiscount = 10.0m;
        ContactPerson = firstName + " " + lastName;
    }

    public override string GetFullName()
    {
        return GroupName;
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Группа: {GroupName}, Участников: {MembersCount}, Тип: {CompanyType}");
        Console.WriteLine($"Корпоративная скидка: {CorporateDiscount}%, Контактное лицо: {ContactPerson}");
    }

    public override void Greet()
    {
        Console.WriteLine($"Здравствуйте! Мы — группа \"{GroupName}\" ({CompanyType})");
    }

    public override void Greet(string greeting)
    {
        Console.WriteLine($"{greeting}! Группа \"{GroupName}\" приветствует вас.");
    }

    public void AddMember(int count)
    {
        MembersCount += count;
        Console.WriteLine($"Добавлено {count} участников. Всего участников: {MembersCount}");
    }
}

public class CustomerCollection<T> where T : Customer
{
    private List<T> _customers = new List<T>();

    public void Add(T customer)
    {
        _customers.Add(customer);
        Console.WriteLine($"Клиент {customer.GetFullName()} добавлен в коллекцию.");
    }

    public void Remove(T customer)
    {
        _customers.Remove(customer);
        Console.WriteLine($"Клиент {customer.GetFullName()} удалён из коллекции.");
    }

    public void DisplayAll()
    {
        Console.WriteLine($"--- В коллекции {_customers.Count} клиент(ов) ---");
        foreach (var customer in _customers)
        {
            customer.ViewProfile();
            Console.WriteLine();
        }
    }

    public void GreetAll()
    {
        foreach (var customer in _customers)
        {
            customer.Greet();
        }
    }
}

Console.WriteLine("=== Демонстрация перегрузки и перекрытия ===");
VipCustomer vip = new VipCustomer(1, "Иван", "Иванов", 35, "ivan@mail.ru", "+7-900-111-22-33",
                                  "Москва, ул. Ленина 1", 1500, "Gold", 15.0);
vip.Greet();
vip.Greet("Добрый день");
vip.ViewProfile();
vip.AddPoints(500);
vip.AddPoints(200, "покупку");
vip.Contact("Email");
vip.UpdateEmail("ivan.vip@mail.ru", true);
vip.AssignPersonalManager("Анна Смирнова");

Console.WriteLine();

RegularCustomer regular = new RegularCustomer(2, "Мария", "Петрова", 28, "maria@mail.ru",
                                              "+7-900-444-55-66", "СПб, Невский пр. 10",
                                              new DateTime(2023, 5, 15), 12, 45000m);
regular.Greet("Здравствуйте");
regular.ViewProfile();
regular.AddOrder(1500m);
regular.UpdateEmail("maria.new@mail.ru", false);

Console.WriteLine();

GroupCustomer group = new GroupCustomer(3, "ООО", "Ромашка", 0, "group@mail.ru",
                                        "+7-495-777-88-99", "Москва, офис 5",
                                        "Корпоративные клиенты", 25, "ООО");
group.Greet("Приветствуем");
group.ViewProfile();
group.AddMember(5);
Console.WriteLine($"Полное имя группы: {group.GetFullName()}");

Console.WriteLine();
Console.WriteLine("=== Демонстрация Generic класса ===");

CustomerCollection<VipCustomer> vipCollection = new CustomerCollection<VipCustomer>();
vipCollection.Add(vip);
vipCollection.Add(new VipCustomer(4, "Петр", "Сидоров", 40, "petr@mail.ru", "+7-900-222-33-44",
                                  "Казань, ул. Баумана 5", 3000, "Platinum", 20.0));
vipCollection.GreetAll();
vipCollection.DisplayAll();

CustomerCollection<Customer> allCustomers = new CustomerCollection<Customer>();
allCustomers.Add(vip);
allCustomers.Add(regular);
allCustomers.Add(group);
allCustomers.DisplayAll();
```
