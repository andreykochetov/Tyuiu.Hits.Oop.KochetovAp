```
using System;

public interface IIdentifiable
{
    int CustomerId { get; set; }
}

public interface IContactable
{
    string Email { get; set; }
    string Phone { get; set; }
    void Contact();
}

public interface ILoyalty
{
    int LoyaltyPoints { get; set; }
    void AddPoints(int points);
}

public abstract class Customer : IIdentifiable, IContactable
{
    public int CustomerId { get; set; }
    public string Name { get; set; }
    public string Email { get; set; }
    public string Phone { get; set; }

    protected Customer(int customerId, string name, string email, string phone)
    {
        CustomerId = customerId;
        Name = name;
        Email = email;
        Phone = phone;
    }

    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"[Базовый] Email обновлен на: {Email}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"--- Профиль клиента ---");
        Console.WriteLine($"ID: {CustomerId} | Имя: {Name}");
        Console.WriteLine($"Email: {Email} | Телефон: {Phone}");
    }

    public void Contact()
    {
        Console.WriteLine($"Связь с {Name} по телефону {Phone}");
    }
}

public class VipCustomer : Customer, ILoyalty
{
    public int LoyaltyPoints { get; set; }
    public string VipLevel { get; set; }

    public VipCustomer(int id, string name, string email, string phone, int points, string level)
        : base(id, name, email, phone)
    {
        LoyaltyPoints = points;
        VipLevel = level;
    }

    public void AddPoints(int points)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"Начислено {points} баллов. Всего: {LoyaltyPoints}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"VIP-уровень: {VipLevel} | Баллы: {LoyaltyPoints}");
    }
}

public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }

    public RegularCustomer(int id, string name, string email, string phone, DateTime regDate)
        : base(id, name, email, phone)
    {
        RegistrationDate = regDate;
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        Console.WriteLine($"[Обычный клиент] Дата обновления: {DateTime.Now:dd.MM.yyyy HH:mm}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Дата регистрации: {RegistrationDate:dd.MM.yyyy}");
    }
}

public class GroupCustomer : Customer, IContactable
{
    public string GroupName { get; set; }

    public GroupCustomer(int id, string name, string email, string phone, string groupName)
        : base(id, name, email, phone)
    {
        GroupName = groupName;
    }

    public override string GetFullName()
    {
        return GroupName;
    }

    string IContactable.Email 
    { 
        get { return Email; } 
        set { Email = value; } 
    }

    string IContactable.Phone 
    { 
        get { return Phone; } 
        set { Phone = value; } 
    }

    void IContactable.Contact()
    {
        Console.WriteLine($"[Явная реализация] Отправка корпоративного запроса в группу {GroupName}");
    }
}

public class CustomerService
{
    private readonly Customer _customer;

    public CustomerService(Customer customer)
    {
        _customer = customer;
    }

    public void ProcessCustomer()
    {
        _customer.ViewProfile();
        _customer.Contact();
        Console.WriteLine();
    }
}

Customer vip = new VipCustomer(1, "Иван Иванов", "ivan@mail.ru", "+7-900-111-22-33", 1500, "Gold");
vip.ViewProfile();
vip.UpdateEmail("ivan.vip@mail.ru");
Console.WriteLine();

Customer regular = new RegularCustomer(2, "Мария Петрова", "maria@mail.ru", "+7-900-444-55-66", new DateTime(2023, 5, 15));
regular.ViewProfile();
regular.UpdateEmail("maria.new@mail.ru");
Console.WriteLine();

GroupCustomer group = new GroupCustomer(3, "ООО Ромашка", "group@mail.ru", "+7-495-777-88-99", "Корпоративные клиенты");
Console.WriteLine($"Имя группы (через GetFullName): {group.GetFullName()}");

IContactable contactGroup = group; 
contactGroup.Contact();
Console.WriteLine();

CustomerService service1 = new CustomerService(vip);
service1.ProcessCustomer();

CustomerService service2 = new CustomerService(regular);
service2.ProcessCustomer();

CustomerService service3 = new CustomerService(group);
service3.ProcessCustomer();

Console.WriteLine("=== ПРОГРАММА ЗАВЕРШЕНА ===");
```
