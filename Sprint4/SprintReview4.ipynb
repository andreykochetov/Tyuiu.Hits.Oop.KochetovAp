```
using System;
using System.Collections.Generic;

public interface ISupplier
{
    string GetName();
    string GetContact();
    decimal GetDeliveryCost();
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; set; }
}

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

public class Enterprise : IEnterprise
{
    private string Name { get; set; }
    private List<ISupplier> suppliers;
    private List<IEmployee> employees;

    public Enterprise(string name)
    {
        Name = name;
        suppliers = new List<ISupplier>();
        employees = new List<IEmployee>();
    }

    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        return suppliers;
    }

    public decimal CalculateTotalExpenses()
    {
        decimal total = 0;
        foreach (var supplier in suppliers)
        {
            total += supplier.GetDeliveryCost();
        }
        foreach (var emp in employees)
        {
            total += emp.GetSalary();
        }
        return total;
    }

    public void PrintReport()
    {
        Console.WriteLine("Список поставщиков:");
        int i = 1;
        foreach (var s in suppliers)
        {
            Console.WriteLine($"{i}. {s.GetName()} — {s.GetContact()} — Стоимость услуги: {s.GetDeliveryCost()}");
            i++;
        }

        Console.WriteLine("\nСписок сотрудников:");
        i = 1;
        foreach (var e in employees)
        {
            Console.WriteLine($"{i}. {e.GetFullName()} — Должность: {e.Position} — Зарплата: {e.GetSalary()}");
            i++;
        }

        Console.WriteLine($"Общие расходы: {CalculateTotalExpenses()}");
    }
}

public class Supplier : ISupplier
{
    public string Name { get; set; }
    public string ContactInfo { get; set; }
    public decimal DeliveryCost { get; set; }

    public Supplier(string name, string contactInfo, decimal deliveryCost)
    {
        Name = name;
        ContactInfo = contactInfo;
        DeliveryCost = deliveryCost;
    }

    public string GetName() => Name;
    public string GetContact() => ContactInfo;
    public decimal GetDeliveryCost() => DeliveryCost;
}

public class Employee : IEmployee
{
    public string FirstName { get; set; }
    public string LastName { get; set; }
    public string Position { get; set; }
    public decimal Salary { get; set; }

    public Employee(string firstName, string lastName, string position, decimal salary)
    {
        FirstName = firstName;
        LastName = lastName;
        Position = position;
        Salary = salary;
    }

    public string GetFullName() => $"{FirstName} {LastName}";
    public decimal GetSalary() => Salary;
}

var enterprise = new Enterprise("ООО Ромашка");

var supplier1 = new Supplier("Логистический центр", "Стоимость услуги", 10000);
var supplier2 = new Supplier("Грузовые перевозки", "Стоимость услуги", 30000);

enterprise.AddSupplier(supplier1);
enterprise.AddSupplier(supplier2);

var emp1 = new Employee("Роман", "Игнатов", "Менеджер", 70000);
var emp2 = new Employee("Таисия", "Землеройка", "Разработчик", 90000);

enterprise.AddEmployee(emp1);
enterprise.AddEmployee(emp2);

enterprise.PrintReport();

Console.WriteLine("\n=== ПРОГРАММА ЗАВЕРШЕНА ===");
```
