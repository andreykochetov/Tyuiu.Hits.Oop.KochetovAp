#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Customer
{
    private int _customerId;
    private string _name;
    private string _email;

    public int CustomerId
    {
        get { return _customerId; }
        set
        {
            if (value <= 0)
                throw new ArgumentException("ID клиента должен быть положительным!");
            _customerId = value;
        }
    }

    public string Name
    {
        get { return _name; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Имя не может быть пустым!");
            _name = value;
        }
    }

    public string Email
    {
        get { return _email; }
        set
        {
            if (string.IsNullOrWhiteSpace(value) || !value.Contains("@"))
                throw new ArgumentException("Некорректный email!");
            _email = value;
        }
    }

    public static int TotalCustomers { get; private set; } = 0;

    public Customer()
    {
        TotalCustomers++;
    }

    public Customer(int id, string name, string email) : this()
    {
        CustomerId = id;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName()
    {
        return Name;
    }

    public virtual void UpdateEmail(string newEmail)
    {
        Email = newEmail;
        Console.WriteLine($"{Name}: email обновлён на {newEmail}");
    }

    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}");
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Email: {Email}");
    }

    public void InteractWith(Customer other)
    {
        Console.WriteLine($"{GetFullName()} взаимодействует с {other.GetFullName()}");
        Console.WriteLine($"  {GetFullName()} -> {Email}");
        Console.WriteLine($"  {other.GetFullName()} -> {other.Email}");
    }

    public static void DisplayTotalCustomers()
    {
        Console.WriteLine($"Всего клиентов: {TotalCustomers}");
    }
}

public class VipCustomer : Customer
{
    private int _loyaltyPoints;

    public int LoyaltyPoints
    {
        get { return _loyaltyPoints; }
        set
        {
            if (value < 0)
                throw new ArgumentException("Баллы не могут быть отрицательными!");
            _loyaltyPoints = value;
        }
    }

    public VipCustomer() : base()
    {
        LoyaltyPoints = 0;
    }

    public VipCustomer(int id, string name, string email, int points)
        : base(id, name, email)
    {
        LoyaltyPoints = points;
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Баллы лояльности: {LoyaltyPoints}");
        Console.WriteLine($"Статус: VIP");
    }
}

public class RegularCustomer : Customer
{
    private DateTime _registrationDate;

    public DateTime RegistrationDate
    {
        get { return _registrationDate; }
        set
        {
            if (value > DateTime.Now)
                throw new ArgumentException("Дата регистрации не может быть в будущем!");
            _registrationDate = value;
        }
    }

    public RegularCustomer() : base()
    {
        RegistrationDate = DateTime.Now;
    }

    public RegularCustomer(int id, string name, string email, DateTime regDate)
        : base(id, name, email)
    {
        RegistrationDate = regDate;
    }

    public override void UpdateEmail(string newEmail)
    {
        base.UpdateEmail(newEmail);
        Console.WriteLine($"Дата последнего обновления email: {DateTime.Now}");
    }

    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Дата регистрации: {RegistrationDate.ToShortDateString()}");
    }
}

public class GroupCustomer : Customer
{
    private string _groupName;

    public string GroupName
    {
        get { return _groupName; }
        set
        {
            if (string.IsNullOrWhiteSpace(value))
                throw new ArgumentException("Название группы не может быть пустым!");
            _groupName = value;
        }
    }

    public GroupCustomer() : base()
    {
        GroupName = "Без группы";
    }

    public GroupCustomer(int id, string name, string email, string group)
        : base(id, name, email)
    {
        GroupName = group;
    }

    public override string GetFullName()
    {
        return GroupName;
    }

    public override void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}");
        Console.WriteLine($"Группа: {GroupName}");
        Console.WriteLine($"Контактное лицо: {Name}");
        Console.WriteLine($"Email: {Email}");
    }
}

VipCustomer vip = new VipCustomer(1, "Алексей", "alex@mail.com", 1500);
RegularCustomer regular = new RegularCustomer(2, "Мария", "maria@mail.com", DateTime.Now.AddDays(-30));
GroupCustomer group = new GroupCustomer(3, "Иван", "group@company.com", "ТехноКомпани");

vip.ViewProfile();
Console.WriteLine();

regular.ViewProfile();
Console.WriteLine();

group.ViewProfile();
Console.WriteLine($"Полное имя: {group.GetFullName()}");
Console.WriteLine();

regular.UpdateEmail("maria.new@mail.com");
Console.WriteLine();

vip.InteractWith(regular);
group.InteractWith(vip);
Console.WriteLine();

Customer.DisplayTotalCustomers();
```
