```
using System;

public class Plant
{
    public string Name { get; set; }
    public double Height { get; set; }

    public Plant( string name, double height)
    {
        Name = name
        Height = height;
    }

    public virtual void Blassom() => Console.WriteLine($"Расстение{Name}цветет");
    
    public void Grow(double g)
    {
        Height += g;
        Console.WriteLine($"{Name} выросло на {g} см. Высота: {Height} см");

    }
}

public class Flower : Plant
{
    public Flower(string name, double height) : base(name, height) { }

    public override void Blassom() => Console.WriteLine($"Цветов{ Name} распускатеся и раудут глаз!");
    public void Wither() => Console.WriteLine($"Цветок {Name} увядает...");

}

public class Tree : Plant
{
    public int Age { get; set; }
    public Tree(string name, double height, int age) : base(name, height) { Age = age;}

    public override void Blassom() => Console.WriteLine($"Дерево {Name} покрывается новыми листьями!" );
    public void GrowUP(int y)
    {
        Age += y;
        Console.WriteLine($"{Name} выросло на {y} лет. Возраст: {Age} лет");

    }
}
Flower rose = new Flower("Роза", 25);
Tree oak = new Tree("Дуб", 150, 10);

rose.Blsassom()
rose.Grow(5);
rose.Wither();

oak.Blassom();
oak.GrowUP(5);

Plant[] plants = { rose, oak };
foreach (Plant p in plants) p.Blassom;

```
