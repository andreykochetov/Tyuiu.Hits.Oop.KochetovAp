<h2 style="color:DodgerBlue">Описание проекта:</h2>



Создать базовый класс Customer в C#, который будет представлять информацию о
клиентах или покупателях. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.



<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

public class Customer
{
    public int CustomerId { get; set; }
    public string Name { get; set; }
    public string Email { get; set; }

    public Customer(int id, string name, string email)
    {
        CustomerId = id;
        Name = name;
        Email = email;
    }

    public virtual string GetFullName() => Name;
    public virtual void UpdateEmail(string newEmail) => Email = newEmail;
    public virtual void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}\nИмя: {Name}\nEmail: {Email}");
    }
}

public class VipCustomer : Customer
{
    public int LoyaltyPoints { get; set; }
    public VipCustomer(int id, string name, string email, int points) : base(id, name, email)
    {
        LoyaltyPoints = points;
    }
    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Баллы: {LoyaltyPoints}\nСтатус: VIP");
    }
}

public class RegularCustomer : Customer
{
    public DateTime RegistrationDate { get; set; }
    public RegularCustomer(int id, string name, string email, DateTime regDate) : base(id, name, email)
    {
        RegistrationDate = regDate;
    }
    public override void ViewProfile()
    {
        base.ViewProfile();
        Console.WriteLine($"Дата регистрации: {RegistrationDate.ToShortDateString()}");
    }
}

public class GroupCustomer : Customer
{
    public string GroupName { get; set; }
    public GroupCustomer(int id, string name, string email, string group) : base(id, name, email)
    {
        GroupName = group;
    }
    public override string GetFullName() => GroupName;
    public override void ViewProfile()
    {
        Console.WriteLine($"ID: {CustomerId}\nГруппа: {GroupName}\nEmail: {Email}");
    }
}

class Program
{
    static void Main()
    {
        VipCustomer vip = new VipCustomer(1, "Алексей", "alex@mail.com", 1500);
        RegularCustomer regular = new RegularCustomer(2, "Мария", "maria@mail.com", DateTime.Now.AddDays(-30));
        GroupCustomer group = new GroupCustomer(3, "Иван", "group@company.com", "ТехноКомпани");

        vip.ViewProfile();
        Console.WriteLine();
        regular.ViewProfile();
        Console.WriteLine();
        group.ViewProfile();
        Console.WriteLine($"Полное имя: {group.GetFullName()}");
    }
}
```
